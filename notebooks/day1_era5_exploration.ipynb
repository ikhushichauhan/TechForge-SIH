```python
import xarray as xr

file_path = "../data/raw/era5_raw.grib"

weather = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "indexpath": ""
    }
)

print("===== WEATHER DATA =====")
print(weather)
```

```python
ssrd = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "filter_by_keys": {
            "shortName": "ssrd"
        },
        "indexpath": ""
    }
)

print("===== SSRD =====")
print(ssrd)
```

```python
print(ssrd["time"].isel(time=slice(0, 3)).values)
print(ssrd["step"].values)
```

```python
ssrd_sample = ssrd["ssrd"].isel(latitude=1, longitude=1)

print("Forecast initialization:")
print(ssrd["time"].isel(time=0).values)

print("\nSteps:")
print(ssrd["step"].values)

print("\nAccumulated SSRD:")
print(ssrd_sample.isel(time=0).values)
```

```python
print("\nNext forecast initialization:")
print(ssrd["time"].isel(time=1).values)

print("\nAccumulated SSRD:")
print(ssrd_sample.isel(time=1).values)
```

```python
strd = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "filter_by_keys": {
            "shortName": "strd"
        },
        "indexpath": ""
    }
)

print("===== STRD =====")
print(strd)
```

```python
ssr = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "filter_by_keys": {
            "shortName": "ssr"
        },
        "indexpath": ""
    }
)

print("===== STRD =====")
print(ssr)
```

```python
str = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "filter_by_keys": {
            "shortName": "str"
        },
        "indexpath": ""
    }
)

print("===== STRD =====")
print(str)
```

```python
fdir = xr.open_dataset(
    file_path,
    engine="cfgrib",
    backend_kwargs={
        "filter_by_keys": {
            "shortName": "fdir"
        },
        "indexpath": ""
    }
)

print("===== STRD =====")
print(fdir)
```

```python
import pandas as pd

# Convert the 2D valid_time coordinate into a 1D time coordinate
ssrd_flat = (
    ssrd
    .stack(sample=("time", "step"))
    .reset_index("sample")
)

strd_flat = (
    strd
    .stack(sample=("time", "step"))
    .reset_index("sample")
)

ssr_flat = (
    ssr
    .stack(sample=("time", "step"))
    .reset_index("sample")
)

str_flat =  (
    str
    .stack(sample=("time", "step"))
    .reset_index("sample")
)

fdir_flat =  (
    fdir
    .stack(sample=("time", "step"))
    .reset_index("sample")
)

# Replace the time coordinate with the actual valid time
ssrd_flat = ssrd_flat.assign_coords(
    valid_time=("sample", ssrd["valid_time"].values.ravel())
)

ssr_flat = ssr_flat.assign_coords(
    valid_time=("sample", ssr["valid_time"].values.ravel())
)

str_flat = str_flat.assign_coords(
    valid_time=("sample", str["valid_time"].values.ravel())
)

strd_flat = strd_flat.assign_coords(
    valid_time=("sample", strd["valid_time"].values.ravel())
)

fdir_flat = fdir_flat.assign_coords(
    valid_time=("sample", fdir["valid_time"].values.ravel())
)

# Remove invalid/duplicate structure if necessary
ssrd_flat = ssrd_flat.swap_dims({"sample": "valid_time"})
ssr_flat = ssr_flat.swap_dims({"sample": "valid_time"})
strd_flat = strd_flat.swap_dims({"sample": "valid_time"})
str_flat = str_flat.swap_dims({"sample": "valid_time"})
fdir_flat = fdir_flat.swap_dims({"sample": "valid_time"})

print(ssrd_flat)
print(ssr_flat)
print(strd_flat)
print(str_flat)
print(fdir_flat)
```

```python
print(
    "Unique timestamps:",
    len(pd.Index(ssrd_flat.valid_time.values).unique())
)

print(
    "Total timestamps:",
    len(ssrd_flat.valid_time)
)

print(
    "Unique timestamps:",
    len(pd.Index(strd_flat.valid_time.values).unique())
)

print(
    "Total timestamps:",
    len(strd_flat.valid_time)
)
```

```python
ssrd_clean = ssrd_flat[["ssrd"]].drop_vars(
    [coord for coord in ["time", "step"] if coord in ssrd_flat.coords]
)

ssr_clean = ssr_flat[["ssr"]].drop_vars(
    [coord for coord in ["time", "step"] if coord in ssr_flat.coords]
)

strd_clean = strd_flat[["strd"]].drop_vars(
    [coord for coord in ["time", "step"] if coord in strd_flat.coords]
)

str_clean = str_flat[["str"]].drop_vars(
    [coord for coord in ["time", "step"] if coord in str_flat.coords]
)

fdir_clean = fdir_flat[["fdir"]].drop_vars(
    [coord for coord in ["time", "step"] if coord in fdir_flat.coords]
)
print("===== SSRD CLEAN =====")
print(ssrd_clean)

print("\n===== SSR CLEAN =====")
print(ssr_clean)

print("\n===== STRD CLEAN =====")
print(strd_clean)

print("\n===== STR CLEAN =====")
print(str_clean)

print("\n===== fdir CLEAN =====")
print(fdir_clean)
```

```python
# Get the timestamps that exist in BOTH datasets
common_times = weather.time.to_index().intersection(
    ssrd_clean.valid_time.to_index()
)

print("Weather timestamps:", len(weather.time))
print("Radiation timestamps:", len(ssrd_clean.valid_time))
print("Common timestamps:", len(common_times))
```

```python
weather_aligned = weather.sel(
    time=common_times
)

ssrd_aligned = ssrd_clean.sel(
    valid_time=weather.time
)

ssr_aligned = ssr_clean.sel(
    valid_time=weather.time
)

strd_aligned = strd_clean.sel(
    valid_time=weather.time
)

str_aligned = str_clean.sel(
    valid_time=weather.time
)

fdir_aligned = fdir_clean.sel(
    valid_time=weather.time
)
```

```python
print(weather_aligned)
print(ssrd_aligned)
print(ssr_aligned)
print(strd_aligned)
print(str_aligned)
print(fdir_aligned)
```

```python
ssrd_final = (
    ssrd_aligned
    .drop_vars("time")
    .rename({"valid_time": "time"})
)

ssr_final = (
    ssr_aligned
    .drop_vars("time")
    .rename({"valid_time": "time"})
)

strd_final = (
    strd_aligned
    .drop_vars("time")
    .rename({"valid_time": "time"})
)

str_final = (
    str_aligned
    .drop_vars("time")
    .rename({"valid_time": "time"})
)

fdir_final = (
    fdir_aligned
    .drop_vars("time")
    .rename({"valid_time": "time"})
)

print("SSRD:")
print(ssrd_final)

print("\nSSR:")
print(ssr_final)

print("\nSTRD:")
print(strd_final)

print("\nSTR:")
print(str_final)

print("\nfdir:")
print(fdir_final)
```

```python
print(
    "Weather == SSRD:",
    weather.time.values.tolist() == ssrd_final.time.values.tolist()
)

print(
    "Weather == SSR:",
    weather.time.values.tolist() == ssr_final.time.values.tolist()
)

print(
    "Weather == STRD:",
    weather.time.values.tolist() == strd_final.time.values.tolist()
)

print(
    "Weather == STR:",
    weather.time.values.tolist() == str_final.time.values.tolist()
)

print(
    "Weather == fdir:",
    weather.time.values.tolist() == fdir_final.time.values.tolist()
)
```

```python
weather_times = weather_aligned.time.to_index()
ssrd_times = ssrd_final.time.to_index()
strd_times = strd_final.time.to_index()

print("Weather:", len(weather_times))
print("SSRD:", len(ssrd_times))
print("STRD:", len(strd_times))

print("\nWeather only vs SSRD:")
print(weather_times.difference(ssrd_times)[:20])

print("\nWeather only vs STRD:")
print(weather_times.difference(strd_times)[:20])

print("\nSSRD only:")
print(ssrd_times.difference(weather_times)[:20])

print("\nSTRD only:")
print(strd_times.difference(weather_times)[:20])

print("\nCommon Weather + SSRD:")
print(len(weather_times.intersection(ssrd_times)))

print("Common Weather + STRD:")
print(len(weather_times.intersection(strd_times)))
```

```python
ssrd_aligned = ssrd_clean.sel(
    valid_time=weather.time
)

ssr_aligned = ssr_clean.sel(
    valid_time=weather.time
)

strd_aligned = strd_clean.sel(
    valid_time=weather.time
)

str_aligned = str_clean.sel(
    valid_time=weather.time
)

fdir_aligned = fdir_clean.sel(
    valid_time=weather.time
)

print("Weather timestamps:", len(weather.time))
print("SSRD aligned:", len(ssrd_aligned.time))
print("SSR aligned:", len(ssr_aligned.time))
print("STRD aligned:", len(strd_aligned.time))
print("STR aligned:", len(str_aligned.time))
print("FDIR aligned:", len(fdir_aligned.time))
```

```python
era5 = xr.merge(
    [
        weather,
        ssrd_final,
        ssr_final,
        strd_final,
        str_final,
        fdir_final
    ],
    compat="override",
    join="exact"
)

print("===== FINAL ERA5 MERGED DATASET =====")
print(era5)
```

```python
import os

output_path = "../data/processed/era5_merged.nc"

# Check whether an old file exists
if os.path.exists(output_path):
    print("Existing era5_merged.nc found.")

    try:
        os.remove(output_path)
        print("Old/broken file deleted.")
    except PermissionError:
        print(
            "ERROR: The file is locked.\n"
            "Restart the VS Code/Jupyter kernel and run this cell again."
        )

# Save the current dataset
if not os.path.exists(output_path):

    era5.to_netcdf(
        output_path,
        engine="netcdf4",
        mode="w"
    )

    print(f"Saved successfully: {output_path}")

else:
    print("File still exists, so it was not overwritten.")
```
